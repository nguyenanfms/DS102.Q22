```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
n_samples = 200

mu1 = np.array([2, 2])
mu2 = np.array([8, 3])
mu3 = np.array([3, 6])
sigma = np.eye(2)

cluster1 = np.random.multivariate_normal(mu1, sigma, n_samples)
cluster2 = np.random.multivariate_normal(mu2, sigma, n_samples)
cluster3 = np.random.multivariate_normal(mu3, sigma, n_samples)

X = np.vstack((cluster1, cluster2, cluster3))


# K-means với thuật toán EM
def kmeans_em(X, K=3, max_iters=100):
    # Khởi tạo ngẫu nhiên K tâm cụm từ tập dữ liệu X
    random_indices = np.random.choice(X.shape[0], K, replace=False)
    centroids = X[random_indices]

    for i in range(max_iters):
        #Gán nhãn cho mỗi điểm dữ liệu dựa trên khoảng cách đến các tâm cụm
        distances = np.linalg.norm(X[:, np.newaxis] - centroids, axis=2)
        labels = np.argmin(distances, axis=1)

        #Cập nhật các tâm cụm bằng cách tính trung bình của các điểm dữ liệu thuộc về mỗi cụm
        new_centroids = np.zeros_like(centroids)
        for k in range(K):
            points_in_cluster = X[labels == k]
            if len(points_in_cluster) > 0:
                new_centroids[k] = points_in_cluster.mean(axis=0)
            else:
                new_centroids[k] = X[np.random.choice(X.shape[0])]

        #Kiểm tra điều kiện hội tụ sớm
        if np.allclose(centroids, new_centroids, atol=1e-4):
            print(f"Thuật toán hội tụ tại vòng lặp thứ: {i}")
            break

        centroids = new_centroids

    return centroids, labels


# Thực thi thuật toán
centroids, labels = kmeans_em(X, K=3)
print("Các tâm cụm tìm được:\n", centroids)


#Trực quan hóa kết quả phân cụm
plt.figure(figsize=(10, 6))

plt.scatter(
    X[:, 0], X[:, 1], c=labels, cmap="viridis", s=30, alpha=0.8, label="Data points"
)

# Vẽ các tâm cụm tìm được bởi K-means
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    c="red",
    marker=".",
    s=200,
    edgecolors="black",
    linewidths=1.5,
    label="K-means Centroids",
)

# Hiển thị nhãn tọa độ của các centroid tìm được
for i, centroid in enumerate(centroids):
    plt.text(
        centroid[0] + 0.15,
        centroid[1] + 0.15,
        f"({centroid[0]:.2f}, {centroid[1]:.2f})",
        fontsize=11,
        fontweight="bold",
        bbox=dict(facecolor="white", alpha=0.7, edgecolor="none"),
    )

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend(loc="upper right")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

# Gaussian Mixture Model 

```python
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
from GaussianMixtureModel import GaussianMixtureModel

path = "cow.jpg"
img = Image.open(path).convert("RGB")
img_np = np.array(img)
h, w, c = img_np.shape 

# Biến đổi thành mảng (N, 3) để GMM học cả 3 màu cùng lúc
X_pixels = img_np.reshape(-1, 3).astype(np.float64)

gmm = GaussianMixtureModel(K=2, max_iters=100)
gmm.fit(X_pixels)
labels = gmm.predict(X_pixels)

mask = labels.reshape(h, w)
bg_cluster_idx = np.argmax(np.bincount(labels))

# Tạo ảnh kết quả từ ảnh màu gốc
filtered_image = img_np.copy()
filtered_image[mask == bg_cluster_idx] = [0, 0, 0]

# Xuất ra file ảnh đã tách nền
output_image = Image.fromarray(filtered_image)
output_image.save("cow_filtered.png")
print("Đã tách nền thành công! Ảnh được lưu với tên: cow_filtered.png")

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

axes[0].imshow(img_np)
axes[0].set_title("1. Original Color Image", fontsize=12)
axes[0].axis("off")

axes[1].imshow(mask, cmap="coolwarm")
axes[1].set_title("2. RGB GMM Mask (No Holes)", fontsize=12)
axes[1].axis("off")

axes[2].imshow(filtered_image)
axes[2].set_title("3. Perfect Background Filtered", fontsize=12)
axes[2].axis("off")

plt.tight_layout()
plt.show()
```
